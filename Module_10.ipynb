# Assignment 1 - Working with NumPy

**Objective:** Understand the basics of NumPy, array creation, and manipulation.

## Step 1: Import NumPy

Before using any NumPy functionality, we need to import the library. The convention is to import it as `np` for shorter, cleaner code.

```python
import numpy as np
```

## 1. 1D Array Operations

We create a 1D array containing integers from 1 to 20 using `np.arange()`, which generates evenly spaced values within a given range (the stop value is exclusive, so we use 21).

```python
arr_1d = np.arange(1, 21)
print("1D Array:", arr_1d)
```

### a. Sum, Mean, Median, and Standard Deviation

- `np.sum()` adds up all elements.
- `np.mean()` calculates the average.
- `np.median()` finds the middle value.
- `np.std()` measures how spread out the values are from the mean.

```python
arr_sum = np.sum(arr_1d)
arr_mean = np.mean(arr_1d)
arr_median = np.median(arr_1d)
arr_std = np.std(arr_1d)

print("Sum:", arr_sum)
print("Mean:", arr_mean)
print("Median:", arr_median)
print("Standard Deviation:", round(arr_std, 4))
```

### b. Indices of Elements Greater Than 10

`np.where()` returns the indices where a given condition is `True`. Here we check which elements are greater than 10.

```python
indices_gt_10 = np.where(arr_1d > 10)[0]
print("Indices of elements > 10:", indices_gt_10)
print("Corresponding values:", arr_1d[indices_gt_10])
```

## 2. 2D Array (4x4) Operations

We create a 4x4 array with numbers 1 to 16 using `np.arange()` combined with `.reshape()`, which reorganizes a 1D array into the specified 2D shape.

```python
arr_2d = np.arange(1, 17).reshape(4, 4)
```

### a. Print the Array

```python
print("2D Array (4x4):")
print(arr_2d)
```

### b. Transpose of the Array

The transpose flips the array over its diagonal, turning rows into columns and vice versa. NumPy's `.T` attribute does this.

```python
transpose_arr = arr_2d.T
print("Transpose:")
print(transpose_arr)
```

### c. Row-wise and Column-wise Sums

Using the `axis` parameter in `np.sum()`:
- `axis=1` sums across each row.
- `axis=0` sums down each column.

```python
row_sums = np.sum(arr_2d, axis=1)
col_sums = np.sum(arr_2d, axis=0)

print("Row-wise sums:", row_sums)
print("Column-wise sums:", col_sums)
```

## 3. Element-wise Operations and Dot Product on 3x3 Arrays

We generate two 3x3 arrays of random integers between 1 and 20 using `np.random.randint()`. A seed is set so the results are reproducible.

```python
np.random.seed(42)
array_a = np.random.randint(1, 21, size=(3, 3))
array_b = np.random.randint(1, 21, size=(3, 3))

print("Array A:")
print(array_a)
print("\nArray B:")
print(array_b)
```

### a. Element-wise Addition, Subtraction, and Multiplication

Standard arithmetic operators on NumPy arrays operate element-by-element (not matrix operations).

```python
addition = array_a + array_b
subtraction = array_a - array_b
multiplication = array_a * array_b

print("Addition:\n", addition)
print("\nSubtraction:\n", subtraction)
print("\nMultiplication (element-wise):\n", multiplication)
```

### b. Dot Product

The dot product (matrix multiplication) is computed using `np.dot()` or the `@` operator, which follows linear algebra rules rather than element-wise multiplication.

```python
dot_product = np.dot(array_a, array_b)
print("Dot Product:")
print(dot_product)
```

## 4. Reshaping and Slicing

We start with a 1D array of size 12 and reshape it into a 3x4 2D array. Then we slice the first two rows and the last two columns.

```python
arr_12 = np.arange(1, 13)
print("Original 1D array:", arr_12)

reshaped = arr_12.reshape(3, 4)
print("\nReshaped (3x4):")
print(reshaped)
```

Slicing syntax: `array[row_start:row_stop, col_start:col_stop]`. We want the first 2 rows (`:2`) and the last 2 columns (`-2:`).

```python
sliced = reshaped[:2, -2:]
print("First two rows, last two columns:")
print(sliced)
```

### Summary — Assignment 1

We practiced creating 1D/2D arrays, computing statistics, boolean indexing, transposing, row/column sums, element-wise arithmetic vs. dot product, and reshaping/slicing.

# Assignment 2 - Working with Pandas

**Objective:** Learn to create and manipulate DataFrames for data analysis.

## Step 1: Import Pandas

We import Pandas as `pd`, the standard convention.

```python
import pandas as pd
```

## 1. Creating the DataFrame

We build a DataFrame from a dictionary, where each key becomes a column name and each list becomes the column's values.

```python
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve'],
    'Age': [24, 27, 22, 32, 29],
    'Department': ['HR', 'Finance', 'IT', 'Marketing', 'HR'],
    'Salary': [45000, 54000, 50000, 62000, 47000]
}

df = pd.DataFrame(data)
df
```

### a. First Five Rows

`.head()` displays the first n rows of the DataFrame (default is 5). Since our dataset only has 5 rows, this shows the entire dataset.

```python
print("First 5 rows:")
print(df.head())
```

### b. Summary Statistics for Age and Salary

`.describe()` gives count, mean, standard deviation, min, max, and quartiles for numeric columns.

```python
summary_stats = df[['Age', 'Salary']].describe()
print(summary_stats)
```

### c. Average Salary in HR Department

We filter rows where `Department == 'HR'`, then take the mean of the `Salary` column for those rows.

```python
hr_avg_salary = df[df['Department'] == 'HR']['Salary'].mean()
print("Average salary in HR department:", hr_avg_salary)
```

## 2. Adding a 'Bonus' Column

We create a new column by multiplying the existing `Salary` column by 0.10 (10%).

```python
df['Bonus'] = df['Salary'] * 0.10
df
```

## 3. Filtering Employees Aged 25 to 30

We use boolean indexing with the `&` operator to combine two conditions: Age >= 25 and Age <= 30.

```python
filtered_df = df[(df['Age'] >= 25) & (df['Age'] <= 30)]
filtered_df
```

## 4. Grouping by Department

`.groupby()` splits the data into groups based on the 'Department' column, and we then calculate the mean salary within each group.

```python
dept_avg_salary = df.groupby('Department')['Salary'].mean()
print(dept_avg_salary)
```

## 5. Sorting by Salary and Saving to CSV

`.sort_values()` sorts the DataFrame by the specified column (ascending by default). We then save the result using `.to_csv()`.

```python
sorted_df = df.sort_values(by='Salary', ascending=True)
print(sorted_df)

sorted_df.to_csv('sorted_by_salary.csv', index=False)
print("\nSaved to sorted_by_salary.csv")
```

### Summary — Assignment 2

We practiced creating a DataFrame, viewing/summarizing data, filtering, adding computed columns, grouping, and sorting/exporting to CSV.

# Assignment 3 - Working with Matplotlib

**Objective:** Practice data visualization techniques for better data representation.

## Step 1: Import Matplotlib

We import `matplotlib.pyplot` as `plt`, the standard convention for plotting.

```python
import matplotlib.pyplot as plt

%matplotlib inline
```

## 1. Line Plot

We plot a simple line chart of `x` vs `y` using `plt.plot()`.

```python
x = [1, 2, 3, 4, 5]
y = [10, 15, 25, 30, 50]
```

### a & b. Plot the Data with Title, Labels, and Grid

- `plt.plot()` draws the line.
- `plt.title()`, `plt.xlabel()`, `plt.ylabel()` add text labels.
- `plt.grid()` adds gridlines for easier reading.
- Markers (`marker='o'`) highlight each data point.

```python
plt.figure(figsize=(7, 5))
plt.plot(x, y, marker='o', color='royalblue', linewidth=2)
plt.title('Simple Line Plot')
plt.xlabel('X Values')
plt.ylabel('Y Values')
plt.grid(True)
plt.show()
```

## 2. Bar Graph - Student Marks

We use `plt.bar()` to represent categorical data (students) against numeric values (marks).

```python
students = ['John', 'Jane', 'Alice', 'Bob']
marks = [75, 85, 60, 90]
```

### a & b. Plot as Bar Graph with Custom Colors and Title

We pass a list of colors to the `color` parameter so each bar has a distinct color.

```python
colors = ['#ff6b6b', '#4ecdc4', '#ffe66d', '#1a936f']

plt.figure(figsize=(7, 5))
plt.bar(students, marks, color=colors)
plt.title('Student Marks in Subject')
plt.xlabel('Students')
plt.ylabel('Marks')
plt.ylim(0, 100)
plt.show()
```

## 3. Pie Chart - Revenue Distribution

We use `plt.pie()` to show the percentage share of revenue by region.

```python
regions = ['North America', 'Europe', 'Asia', 'Others']
revenue = [45, 25, 20, 10]
```

### a & b. Pie Chart with Labels, Highlighting the Highest Region

We use the `explode` parameter to visually "pull out" the slice with the highest revenue (North America), making it stand out from the rest of the pie.

```python
explode = [0.1 if r == max(revenue) else 0 for r in revenue]

plt.figure(figsize=(7, 7))
plt.pie(
    revenue,
    labels=regions,
    autopct='%1.1f%%',
    explode=explode,
    startangle=90,
    colors=['#ff9f43', '#54a0ff', '#5f27cd', '#c8d6e5']
)
plt.title("Company Revenue Distribution by Region")
plt.axis('equal')
plt.show()
```

## 4. Histogram - Frequency Distribution

We generate 1000 random integers between 1 and 100 using `np.random.randint()`, then use `plt.hist()` to visualize their frequency distribution across bins.

```python
np.random.seed(42)
random_data = np.random.randint(1, 101, size=1000)

plt.figure(figsize=(8, 5))
plt.hist(random_data, bins=20, color='mediumseagreen', edgecolor='black')
plt.title('Frequency Distribution of Random Integers (1-100)')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.grid(axis='y', alpha=0.5)
plt.show()
```

### Summary — Assignment 3

We practiced line plots, bar graphs, pie charts (with an exploded highlight slice), and histograms for visualizing distributions.
